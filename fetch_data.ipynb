```python
# ============================================================
# LOAD DATA LOKASI
# ============================================================

filename = "/content/drive/MyDrive/puskesmas-rute/puskesmas_surabaya.csv"

df_lokasi = pd.read_csv(filename)

display(
    df_lokasi[
        [
            "Input Nama",
            "Latitude",
            "Longitude",
            "Kecamatan"
        ]
    ]
)

print("Jumlah lokasi:", len(df_lokasi))
```

```python
# ============================================================
# VALIDASI KOORDINAT
# ============================================================

assert df_lokasi["Latitude"].notna().all(), \
    "Masih ada latitude kosong."

assert df_lokasi["Longitude"].notna().all(), \
    "Masih ada longitude kosong."

print("✅ Semua koordinat tersedia")
```

```python
# ============================================================
# ATRIBUT OSM
# ============================================================

extra_tags = [
    "surface",
    "smoothness",

    "lanes",
    "maxspeed",

    "width",
    "est_width",

    "service",
    "access",
    "motor_vehicle",

    "junction",
    "bridge",
    "tunnel"
]

ox.settings.useful_tags_way = list(
    dict.fromkeys(
        ox.settings.useful_tags_way + extra_tags
    )
)

ox.settings.use_cache = True

print("Tag yang disimpan:")
print(ox.settings.useful_tags_way)
```

```python
# ============================================================
# MEMBUAT BOUNDING BOX DARI SELURUH TITIK
# + MARGIN 2 KM
# ============================================================

gdf_lokasi = gpd.GeoDataFrame(
    df_lokasi.copy(),
    geometry=gpd.points_from_xy(
        df_lokasi["Longitude"],
        df_lokasi["Latitude"]
    ),
    crs="EPSG:4326"
)

# Cari sistem koordinat UTM lokal
utm_crs = gdf_lokasi.estimate_utm_crs()

gdf_utm = gdf_lokasi.to_crs(utm_crs)

# Ambil batas titik
minx, miny, maxx, maxy = gdf_utm.total_bounds

# Tambahkan margin 2 km
margin_m = 2000

bbox_utm = box(
    minx - margin_m,
    miny - margin_m,
    maxx + margin_m,
    maxy + margin_m
)

# Balik lagi ke latitude-longitude
bbox_wgs = (
    gpd.GeoSeries(
        [bbox_utm],
        crs=utm_crs
    )
    .to_crs("EPSG:4326")
    .iloc[0]
)

left, bottom, right, top = bbox_wgs.bounds

bbox = (
    left,
    bottom,
    right,
    top
)

print("Bounding box:")
print("West :", left)
print("South:", bottom)
print("East :", right)
print("North:", top)
```

```python
# ============================================================
# DEFINISI PATH GRAPH
# ============================================================
graph_path = "/content/drive/MyDrive/puskesmas-rute/surabaya_drive.graphml"

# ============================================================
# CEK APAKAH GRAPH SUDAH TERSEDIA
# ============================================================

if os.path.exists(graph_path):

    print("Ģ Graph ditemukan di Google Drive")
    print("Memuat graph yang sudah disimpan...")

    G = ox.io.load_graphml(
        filepath=graph_path
    )

    print("✅ Graph berhasil dimuat tanpa fetching ulang")

else:

    print("ጁ Graph belum tersedia")
    print("Mengambil jaringan jalan dari OpenStreetMap...")

    G = ox.graph.graph_from_bbox(
        bbox,
        network_type="drive_service",
        simplify=True,
        retain_all=False,
        truncate_by_edge=True
    )

    print("✅ Jaringan jalan berhasil diambil")

    # Tambahkan estimasi kecepatan
    G = ox.routing.add_edge_speeds(G)

    # Tambahkan estimasi waktu perjalanan
    G = ox.routing.add_edge_travel_times(G)

    # Simpan graph ke Google Drive
    ox.io.save_graphml(
        G,
        filepath=graph_path
    )

    print("✅ Graph berhasil disimpan")

# ============================================================
# VALIDASI ATRIBUT SETELAH LOAD
# ============================================================

if not all(
    "speed_kph" in d
    for _, _, _, d in G.edges(keys=True, data=True)
):
    G = ox.routing.add_edge_speeds(G)

if not all(
    "travel_time" in d
    for _, _, _, d in G.edges(keys=True, data=True)
):
    G = ox.routing.add_edge_travel_times(G)

    ox.io.save_graphml(
        G,
        filepath=graph_path
    )

# ============================================================
# INFORMASI GRAPH
# ============================================================

print("\n===== INFORMASI GRAPH =====")

print("Jumlah node:", len(G.nodes))
print("Jumlah edge:", len(G.edges))
print("File graph:", graph_path)
print("Ukuran file (MB):",
      round(os.path.getsize(graph_path) / 1024**2, 2))
```

```python
# ============================================================
# GRAPH → GEODATAFRAME
# ============================================================

nodes, edges = ox.convert.graph_to_gdfs(G)

print("Jumlah ruas:", len(edges))

kolom = [
    "name",

    "highway",
    "length",

    "speed_kph",
    "travel_time",

    "oneway",

    "service",

    "lanes",
    "maxspeed",

    "width",
    "surface"
]

kolom_tersedia = [
    c
    for c in kolom
    if c in edges.columns
]

display(
    edges[kolom_tersedia].head(30)
)
```

```python
# ============================================================
# PENGECEKAN MISSING VALUE PADA RUAS JALAN (EDGES)
# ============================================================

kolom_target = ['length', 'speed_kph', 'travel_time', 'highway', 'oneway']

# Memastikan kolom benar-benar ada di dalam DataFrame
kolom_ada = [col for col in kolom_target if col in edges.columns]

# Hitung jumlah missing value
missing_count = edges[kolom_ada].isna().sum()

# Hitung persentase missing value
missing_percentage = (edges[kolom_ada].isna().sum() / len(edges)) * 100

# Gabungkan ke dalam sebuah ringkasan DataFrame
summary_missing = pd.DataFrame({
    'Jumlah Missing': missing_count,
    'Persentase (%)': missing_percentage.round(2)
})

display(summary_missing)
```

```python
# ============================================================
# DISTRIBUSI HIGHWAY
# ============================================================

def flatten_highway(value):

    if isinstance(value, list):
        return value

    return [value]


hasil = []

for value in edges["highway"].dropna():

    hasil.extend(
        flatten_highway(value)
    )


highway_counts = (
    pd.Series(hasil)
    .value_counts()
    .reset_index()
)

highway_counts.columns = [
    "highway",
    "jumlah_ruas"
]

display(highway_counts)
```

```python
# ============================================================
# PERSENTASE DISTRIBUSI HIGHWAY
# ============================================================

# Hitung total ruas jalan
total_ruas = highway_counts["jumlah_ruas"].sum()

# Hitung persentase
highway_counts["persentase (%)"] = (highway_counts["jumlah_ruas"] / total_ruas) * 100

# Format tampilan persentase agar lebih rapi
highway_counts["persentase (%)"] = highway_counts["persentase (%)"].round(2)

display(highway_counts)
print(f"Total keseluruhan ruas jalan: {total_ruas}")
```

```python
# ============================================================
# DIJKSTRA - MATRIKS JARAK DAN WAKTU
# ============================================================

import numpy as np
import pandas as pd
import networkx as nx
import osmnx as ox
import os

# Data lokasi yang sudah dibaca sebelumnya
df_titik = df_lokasi.copy().reset_index(drop=True)

nama_lokasi = df_titik["Input Nama"].tolist()
```

```python
# ============================================================
# 1. SNAP KOORDINAT KE NODE JARINGAN JALAN
# ============================================================

node_terdekat, jarak_snap = ox.distance.nearest_nodes(
    G,
    X=df_titik["Longitude"].to_numpy(),
    Y=df_titik["Latitude"].to_numpy(),
    return_dist=True
)

df_titik["osm_node"] = node_terdekat
df_titik["jarak_snap_m"] = jarak_snap

display(
    df_titik[
        ["Input Nama", "osm_node", "jarak_snap_m"]
    ]
)

print("Jarak snap terjauh:",
      round(df_titik["jarak_snap_m"].max(), 2),
      "meter")

# ============================================================
# 2. DIJKSTRA
# ============================================================

nodes = df_titik["osm_node"].tolist()

n = len(nodes)

matrix_length = np.full((n, n), np.inf)
matrix_time = np.full((n, n), np.inf)

print(f"Menghitung Dijkstra untuk {n} lokasi...")

for i, origin in enumerate(nodes):

    # Jalur dengan jarak terpendek
    jarak = nx.single_source_dijkstra_path_length(
        G,
        source=origin,
        weight="length"
    )

    # Jalur dengan waktu tercepat
    waktu = nx.single_source_dijkstra_path_length(
        G,
        source=origin,
        weight="travel_time"
    )

    for j, destination in enumerate(nodes):

        matrix_length[i, j] = jarak.get(
            destination, np.inf
        )

        matrix_time[i, j] = waktu.get(
            destination, np.inf
        )

    print(f"✅ {i+1}/{n}: {nama_lokasi[i]}")
```

```python
# ============================================================
# 1. CEK KOMPONEN JARINGAN JALAN
# ============================================================

komponen = list(nx.strongly_connected_components(G))
komponen_utama = max(komponen, key=len)

print("Jumlah komponen:", len(komponen))
print("Jumlah node awal:", len(G.nodes))
print("Node komponen utama:", len(komponen_utama))

# Ambil jaringan yang bisa dilalui dua arah
G_rute = G.subgraph(komponen_utama).copy()

print("Node graph rute:", len(G_rute.nodes))
print("Edge graph rute:", len(G_rute.edges))

# ============================================================
# 2. CEK NODE TERDEKAT SETIAP PUSKESMAS
# ============================================================

df_titik = df_lokasi.copy().reset_index(drop=True)

df_titik["Latitude"] = pd.to_numeric(df_titik["Latitude"])
df_titik["Longitude"] = pd.to_numeric(df_titik["Longitude"])

if df_titik[["Latitude", "Longitude"]].isna().any().any():
    raise ValueError("Ada koordinat yang kosong.")

if not df_titik["Latitude"].between(-90, 90).all():
    raise ValueError("Ada latitude yang tidak valid.")

if not df_titik["Longitude"].between(-180, 180).all():
    raise ValueError("Ada longitude yang tidak valid.")

if df_titik["Input Nama"].duplicated().any():
    raise ValueError("Ada nama lokasi yang duplikat.")

lon = df_titik["Longitude"].to_numpy()
lat = df_titik["Latitude"].to_numpy()

# Node terdekat dari graph awal
node_awal, jarak_awal = ox.distance.nearest_nodes(
    G,
    X=lon,
    Y=lat,
    return_dist=True
)

# Cari ulang pada jaringan yang saling terhubung
node_rute, jarak_rute = ox.distance.nearest_nodes(
    G_rute,
    X=lon,
    Y=lat,
    return_dist=True
)

df_titik["node_awal"] = node_awal
df_titik["node_rute"] = node_rute
df_titik["jarak_awal_m"] = jarak_awal
df_titik["jarak_snap_m"] = jarak_rute

df_titik["node_berubah"] = (
    df_titik["node_awal"] != df_titik["node_rute"]
)

print("\nLokasi yang mengalami perubahan node:")

display(
    df_titik.loc[
        df_titik["node_berubah"],
        [
            "Input Nama",
            "node_awal",
            "node_rute",
            "jarak_awal_m",
            "jarak_snap_m"
        ]
    ]
)

# ============================================================
# 3. VALIDASI JARAK KE NODE JALAN
# ============================================================

# Batas pemeriksaan sementara, bukan standar resmi
batas_snap = 300

lokasi_jauh = df_titik[
    df_titik["jarak_snap_m"] > batas_snap
]

print("\nJarak snap terjauh:",
      round(df_titik["jarak_snap_m"].max(), 2), "meter")

if not lokasi_jauh.empty:

    print("\nLokasi yang perlu dicek ulang:")

    display(
        lokasi_jauh[
            ["Input Nama", "jarak_snap_m", "node_rute"]
        ]
    )

    raise ValueError(
        "Ada lokasi yang terlalu jauh dari jaringan utama. "
        "Periksa koordinat atau cakupan graph sebelum lanjut."
    )

print("Semua lokasi berada dalam batas jarak snap.")
```

```python
# ============================================================
# 4. HITUNG MATRIKS DIJKSTRA
# ============================================================

nama_lokasi = df_titik["Input Nama"].tolist()
node_lokasi = df_titik["node_rute"].tolist()

n = len(node_lokasi)

matrix_length = np.full((n, n), np.inf)
matrix_time = np.full((n, n), np.inf)

for i, asal in enumerate(node_lokasi):

    hasil_jarak = nx.single_source_dijkstra_path_length(
        G_rute,
        source=asal,
        weight="length"
    )

    hasil_waktu = nx.single_source_dijkstra_path_length(
        G_rute,
        source=asal,
        weight="travel_time"
    )

    for j, tujuan in enumerate(node_lokasi):

        matrix_length[i, j] = hasil_jarak.get(
            tujuan, np.inf
        )

        matrix_time[i, j] = hasil_waktu.get(
            tujuan, np.inf
        )

    print(f"{i+1}/{n} selesai: {nama_lokasi[i]}")

# ============================================================
# 5. CEK HASIL DIJKSTRA
# ============================================================

cek_jarak = np.isfinite(matrix_length).all()
cek_waktu = np.isfinite(matrix_time).all()

print("\nSemua jarak terhubung:", cek_jarak)
print("Semua waktu terhubung:", cek_waktu)

if not (cek_jarak and cek_waktu):
    raise ValueError("Masih ada pasangan yang tidak terhubung.")

# Cek lokasi yang jatuh pada node yang sama
duplikat_node = df_titik[
    df_titik.duplicated("node_rute", keep=False)
]

if not duplikat_node.empty:
    print("\nAda lokasi berbeda pada node jalan yang sama:")
    display(
        duplikat_node[
            ["Input Nama", "node_rute", "jarak_snap_m"]
        ]
    )

# ============================================================
# 6. BUAT DATAFRAME
# ============================================================

df_length = pd.DataFrame(
    matrix_length,
    index=nama_lokasi,
    columns=nama_lokasi
)

df_time = pd.DataFrame(
    matrix_time,
    index=nama_lokasi,
    columns=nama_lokasi
)

print("\nMatriks jarak (meter):")
display(df_length.iloc[:5, :5])

print("\nMatriks waktu (detik):")
display(df_time.iloc[:5, :5])

# ============================================================
# 7. SIMPAN HASIL KE GOOGLE DRIVE
# ============================================================

os.makedirs(folder_graph, exist_ok=True)

ox.io.save_graphml(
    G_rute,
    filepath=graph_routing_path
)

print("Graph routing berhasil disimpan.")

df_titik.to_csv(
    os.path.join(folder_graph, "lokasi_osm_nodes.csv"),
    index=False
)

df_length.to_csv(
    os.path.join(folder_graph, "matrix_length.csv")
)

df_time.to_csv(
    os.path.join(folder_graph, "matrix_time.csv")
)

np.save(
    os.path.join(folder_graph, "matrix_length.npy"),
    matrix_length
)

np.save(
    os.path.join(folder_graph, "matrix_time.npy"),
    matrix_time
)

print("\nGraph dan matriks berhasil disimpan.")
```
