```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score, calinski_harabasz_score
from sklearn.manifold import MDS
```

```python
folder = "/content/drive/MyDrive/puskesmas-rute"
os.makedirs(folder, exist_ok=True)
```

```python
# Jumlah cluster ditetapkan
k = 11

# Batas jumlah anggota
maks_anggota = 12

print("Jumlah cluster:", k)
print("Maksimal anggota per cluster:", maks_anggota)
```

```python
df_titik = pd.read_csv(
    os.path.join(folder, "lokasi_osm_nodes.csv")
)

matrix_length = np.load(
    os.path.join(folder, "matrix_length.npy")
)

matrix_time = np.load(
    os.path.join(folder, "matrix_time.npy")
)

# Rapikan nama lokasi
df_titik["Input Nama"] = (
    df_titik["Input Nama"].str.strip()
)

# Periksa ukuran data
assert matrix_length.shape == (len(df_titik), len(df_titik))
assert matrix_time.shape == matrix_length.shape
assert np.isfinite(matrix_length).all()
assert np.isfinite(matrix_time).all()
assert df_titik["Input Nama"].is_unique

# Pisahkan Dinkes dari puskesmas
is_depot = df_titik["Input Nama"].str.contains(
    "Dinas Kesehatan|Dinkes",
    case=False,
    na=False
)

assert is_depot.sum() == 1

idx_depot = np.where(is_depot)[0][0]
idx_puskesmas = np.where(~is_depot)[0]

df_puskesmas = df_titik.iloc[idx_puskesmas].copy()
df_puskesmas = df_puskesmas.reset_index(drop=True)

# Ambil jarak antarpuskesmas
D_arah = matrix_length[
    np.ix_(idx_puskesmas, idx_puskesmas)
]

# Jarak pergi dan pulang dirata-ratakan
D = (D_arah + D_arah.T) / 2

D = D.astype(float)

np.fill_diagonal(D, 0)

# Periksa matriks
assert len(df_puskesmas) == 63
assert D.shape == (63, 63)
assert np.isfinite(D).all()
assert np.allclose(D, D.T)
assert (D >= 0).all()
assert not np.any(np.triu(D == 0, k=1))

print("Jumlah puskesmas:", len(df_puskesmas))
print("Ukuran matriks jarak:", D.shape)
print("Ukuran matriks waktu:", matrix_time.shape)
print("Depot:", df_titik.iloc[idx_depot]["Input Nama"])
```

```python
# Mengubah meter ke kilometer
D_km = D / 1000

# Representasi jarak jaringan jalan ke ruang fitur
model_mds = MDS(
    n_components=5,
    dissimilarity="precomputed",
    n_init=1,
    max_iter=300,
    random_state=42
)

X_mds = model_mds.fit_transform(D_km.copy())

print("Matriks untuk clustering:", D.shape)
print("Data untuk Calinski-Harabasz:", X_mds.shape)
```

```python
#AGGLOMERATIVE

def agglomerative(D, k, linkage="average"):

    model = AgglomerativeClustering(
        n_clusters=k,
        metric="precomputed",
        linkage=linkage
    )

    labels = model.fit_predict(D.copy())

    return labels
```

```python
#K-MEDOIDS

def k_medoids(D, k, seed=42, max_iter=100):

    rng = np.random.default_rng(seed)

    n = len(D)

    # Pilih medoid awal secara acak
    medoids = rng.choice(
        n,
        size=k,
        replace=False
    )

    for iterasi in range(max_iter):

        # Jarak setiap puskesmas ke medoid
        jarak_medoid = D[:, medoids]

        # Tentukan cluster terdekat
        labels = np.argmin(jarak_medoid, axis=1)

        medoids_baru = []

        for c in range(k):

            anggota = np.where(labels == c)[0]

            # Semua cluster harus punya anggota
            if len(anggota) == 0:
                raise ValueError("Ada cluster K-Medoids yang kosong.")

            # Jarak antaranggota cluster
            jarak_anggota = D[
                np.ix_(anggota, anggota)
            ]

            # Cari anggota dengan total jarak terkecil
            total_jarak = jarak_anggota.sum(axis=1)

            posisi_terbaik = np.argmin(total_jarak)

            medoid_baru = anggota[posisi_terbaik]

            medoids_baru.append(medoid_baru)

        medoids_baru = np.array(medoids_baru)

        # Berhenti jika medoid tidak berubah
        if np.array_equal(medoids, medoids_baru):
            break

        medoids = medoids_baru

    # Hasil pembagian cluster terakhir
    labels = np.argmin(D[:, medoids], axis=1)

    return labels, medoids
```

```python
def evaluasi_cluster_baru(labels):

    labels = np.array(labels)

    jumlah_cluster = len(np.unique(labels))

    # Silhouette menggunakan matriks jarak jalan
    silhouette = silhouette_score(
        D,
        labels,
        metric="precomputed"
    )

    # Calinski-Harabasz menggunakan hasil MDS
    calinski = calinski_harabasz_score(
        X_mds,
        labels
    )

    # Hitung jumlah anggota tiap cluster
    ukuran_cluster = pd.Series(labels).value_counts()

    anggota_terbesar = int(ukuran_cluster.max())
    anggota_terkecil = int(ukuran_cluster.min())

    memenuhi_batas = anggota_terbesar <= maks_anggota

    return {
        "jumlah_cluster": jumlah_cluster,
        "silhouette": silhouette,
        "calinski_harabasz": calinski,
        "anggota_terbesar": anggota_terbesar,
        "anggota_terkecil": anggota_terkecil,
        "memenuhi_batas": memenuhi_batas
    }
```

```python
hasil_evaluasi_k11 = []

# AGGLOMERATIVE CLUSTERING
for linkage in ["average", "complete"]:

    labels = agglomerative(
        D,
        k=k,
        linkage=linkage
    )

    nilai = evaluasi_cluster_baru(labels)

    hasil_evaluasi_k11.append({
        "metode": "Agglomerative",
        "k": k,
        "linkage": linkage,
        "seed": np.nan,
        **nilai
    })


# K-MEDOIDS
for seed in range(42, 52):

    labels, medoids = k_medoids(
        D,
        k=k,
        seed=seed
    )

    nilai = evaluasi_cluster_baru(labels)

    hasil_evaluasi_k11.append({
        "metode": "K-Medoids",
        "k": k,
        "linkage": "-",
        "seed": seed,
        **nilai
    })


df_evaluasi_k11 = pd.DataFrame(hasil_evaluasi_k11)

print("Jumlah percobaan:", len(df_evaluasi_k11))

print("\nHasil evaluasi keseluruhan:")
display(df_evaluasi_k11.round(4))
```

```python
df_valid_k11 = df_evaluasi_k11[
    (df_evaluasi_k11["memenuhi_batas"]) &
    (df_evaluasi_k11["jumlah_cluster"] == 11)
].copy()

print("Jumlah konfigurasi yang valid:", len(df_valid_k11))

hasil_terbaik_k11 = []

for metode in ["Agglomerative", "K-Medoids"]:

    data = df_valid_k11[
        df_valid_k11["metode"] == metode
    ].copy()

    if data.empty:
        raise ValueError(
            f"Tidak ada hasil {metode} dengan "
            "11 cluster dan maksimal 12 anggota."
        )

    data = data.sort_values(
        ["silhouette", "calinski_harabasz"],
        ascending=[False, False]
    )

    terbaik = data.iloc[0].to_dict()

    hasil_terbaik_k11.append(terbaik)


df_terbaik_k11 = pd.DataFrame(hasil_terbaik_k11)

print("Konfigurasi terbaik masing-masing metode:")

display(
    df_terbaik_k11[
        [
            "metode",
            "k",
            "linkage",
            "seed",
            "silhouette",
            "calinski_harabasz",
            "anggota_terbesar",
            "anggota_terkecil"
        ]
    ].round(4)
)
```

```python
kelompok_model = {}

for _, baris in df_terbaik_k11.iterrows():

    metode = baris["metode"]

    if metode == "Agglomerative":

        labels = agglomerative(
            D,
            k=k,
            linkage=baris["linkage"]
        )

    else:

        labels, medoids = k_medoids(
            D,
            k=k,
            seed=int(baris["seed"])
        )

    # Cek hasil akhir
    assert len(labels) == 63
    assert len(np.unique(labels)) == 11
    assert np.all(labels >= 0)
    assert pd.Series(labels).value_counts().max() <= 12

    kelompok_model[metode] = labels

    print("\nMetode:", metode)
    print("Jumlah cluster:", len(np.unique(labels)))

    print("Jumlah anggota tiap cluster:")

    display(
        pd.Series(labels)
        .value_counts()
        .sort_index()
        .rename("jumlah_anggota")
        .to_frame()
    )
```

```python
# Baca hasil Dijkstra
df_titik = pd.read_csv(
    os.path.join(folder, "lokasi_osm_nodes.csv")
)

matrix_length = np.load(
    os.path.join(folder, "matrix_length.npy")
)

if matrix_length.shape != (len(df_titik), len(df_titik)):
    raise ValueError("Ukuran matriks dan jumlah lokasi tidak sama.")

if not np.isfinite(matrix_length).all():
    raise ValueError("Masih ada nilai inf atau NaN pada matriks jarak.")

if not np.allclose(np.diag(matrix_length), 0):
    raise ValueError("Diagonal matriks jarak harus nol.")

# Pisahkan Dinkes dari data clustering
is_depot = df_titik["Input Nama"].str.contains(
    "Dinas Kesehatan|Dinkes",
    case=False,
    na=False
)

if is_depot.sum() != 1:
    raise ValueError("Periksa baris Dinkes pada data lokasi.")

idx = np.flatnonzero(~is_depot.to_numpy())

df_puskesmas = df_titik.iloc[idx].copy()
df_puskesmas = df_puskesmas.reset_index(drop=True)

# Matriks jarak antarpuskesmas
D_arah = matrix_length[np.ix_(idx, idx)].astype(float)

# Rata-rata jarak pergi dan pulang
D_master = (D_arah + D_arah.T) / 2
np.fill_diagonal(D_master, 0)

if not np.isfinite(D_master).all():
    raise ValueError("Ada jarak yang tidak valid.")

if (D_master < 0).any():
    raise ValueError("Ada jarak negatif.")

if not np.allclose(D_master, D_master.T):
    raise ValueError("Matriks belum simetris.")

if np.any(np.triu(D_master == 0, k=1)):
    raise ValueError("Ada dua lokasi berbeda dengan jarak nol.")

# Simpan matriks dan urutan data puskesmas
np.save(
    os.path.join(folder, "matrix_hdbscan.npy"),
    D_master
)

df_puskesmas.to_csv(
    os.path.join(folder, "puskesmas_hdbscan.csv"),
    index=False
)

# Matriks utama tidak boleh berubah saat fitting
D_master.setflags(write=False)

print("Jumlah puskesmas:", len(df_puskesmas))
print("Ukuran matriks:", D_master.shape)
print("Diagonal maksimum:", np.max(np.abs(np.diag(D_master))))
```

```python
def evaluasi_cluster(labels):

    labels = np.asarray(labels, dtype=int)

    anggota = labels != -1
    labels_anggota = labels[anggota]

    n_cluster = len(np.unique(labels_anggota))
    n_noise = int((~anggota).sum())

    if n_cluster:
        cluster_terbesar = int(
            pd.Series(labels_anggota).value_counts().max()
        )
    else:
        cluster_terbesar = 0

    hasil = {
        "Jumlah Cluster": n_cluster,
        "Noise": n_noise,
        "Noise (%)": round(100 * n_noise / len(labels), 2),
        "Cluster Terbesar": cluster_terbesar,
        "DBCV": np.nan,
        "Silhouette": np.nan,
        "Error DBCV": "",
        "Error Silhouette": ""
    }

    if n_cluster < 2:
        return hasil

    # Silhouette untuk anggota cluster saja
    try:
        D_anggota = D_master[
            np.ix_(anggota, anggota)
        ].copy()

        np.fill_diagonal(D_anggota, 0)

        if n_cluster < len(labels_anggota):
            hasil["Silhouette"] = float(
                silhouette_score(
                    D_anggota,
                    labels_anggota,
                    metric="precomputed"
                )
            )

    except Exception as e:
        hasil["Error Silhouette"] = str(e)

    # DBCV menggunakan seluruh data termasuk noise
    try:
        hasil["DBCV"] = float(
            validity_index(
                D_master.copy(),
                labels.astype(np.int64),
                metric="precomputed",
                d=2
            )
        )

    except Exception as e:
        hasil["Error DBCV"] = str(e)

    return hasil
```

```python
min_cluster_size_list = [3, 4, 5, 6, 7, 8, 10]
min_samples_list = [1, 2, 3, 4, 5]
method_list = ["eom", "leaf"]

kombinasi = [
    (mcs, ms, metode)
    for mcs, ms, metode in product(
        min_cluster_size_list,
        min_samples_list,
        method_list
    )
    if ms <= mcs
]

hasil = []

for no, (mcs, ms, metode) in enumerate(kombinasi, start=1):

    baris = {
        "min_cluster_size": mcs,
        "min_samples": ms,
        "method": metode
    }

    try:
        model = HDBSCAN(
            min_cluster_size=mcs,
            min_samples=ms,
            cluster_selection_method=metode,
            metric="precomputed",
            algorithm="brute",
            copy=True
        )

        label = model.fit_predict(D_master.copy())

        baris.update(evaluasi_cluster(label))
        baris["Status"] = "selesai"
        baris["Error Fitting"] = ""

    except Exception as e:
        baris["Status"] = "gagal fitting"
        baris["Error Fitting"] = str(e)

    hasil.append(baris)

    print(
        f"Proses tuning: {no}/{len(kombinasi)}",
        end="\r"
    )

df_tuning = pd.DataFrame(hasil)

for kolom in ["DBCV", "Silhouette"]:
    df_tuning[kolom] = pd.to_numeric(
        df_tuning[kolom],
        errors="coerce"
    ).replace([np.inf, -np.inf], np.nan)

# Simpan semua hasil tuning
df_tuning.to_csv(
    os.path.join(folder, "hasil_tuning_hdbscan.csv"),
    index=False
)

# Ambil konfigurasi dengan metrik lengkap
df_valid = df_tuning.loc[
    (df_tuning["Jumlah Cluster"] >= 2) &
    df_tuning["DBCV"].notna() &
    df_tuning["Silhouette"].notna()
].copy()

df_valid = df_valid.sort_values(
    ["DBCV", "Silhouette", "Noise (%)"],
    ascending=[False, False, True]
).reset_index(drop=True)

print("\nTotal kombinasi:", len(kombinasi))
print("Berhasil fitting:",
      (df_tuning["Status"] == "selesai").sum())
print("Dapat dievaluasi:", len(df_valid))
print("Gagal fitting:",
      (df_tuning["Status"] == "gagal fitting").sum())

print("Diagonal setelah tuning:",
      np.max(np.abs(np.diag(D_master))))

display(df_valid.head(15).round(4))

# Tampilkan error jika ada
masalah = df_tuning[
    (df_tuning["Status"] != "selesai") |
    (df_tuning["Error DBCV"].fillna("") != "") |
    (df_tuning["Error Silhouette"].fillna("") != "")
]

if not masalah.empty:
    display(masalah.head(20))
```

```python
import os
import numpy as np
import pandas as pd

folder = "/content/drive/MyDrive/puskesmas-rute"

df_rute = pd.read_csv(
    os.path.join(folder, "hasil_hdbscan_final.csv")
)

df_urutan = pd.read_csv(
    os.path.join(folder, "puskesmas_hdbscan.csv")
)

D = np.load(
    os.path.join(folder, "matrix_hdbscan.npy")
)

# Pastikan urutan puskesmas sesuai dengan matriks
if not df_rute["Input Nama"].equals(df_urutan["Input Nama"]):
    raise ValueError("Urutan nama puskesmas tidak sesuai matriks.")

if D.shape != (len(df_rute), len(df_rute)):
    raise ValueError("Ukuran matriks tidak sesuai jumlah lokasi.")

# Simpan label HDBSCAN asli
df_rute["cluster_hdbscan"] = df_rute["cluster"].astype(int)

# Kolom kelompok untuk distribusi
df_rute["kelompok_rute"] = df_rute["cluster_hdbscan"]

cluster_awal = sorted(
    df_rute.loc[
        df_rute["cluster_hdbscan"] != -1,
        "cluster_hdbscan"
    ].unique()
)

noise_idx = np.flatnonzero(
    df_rute["cluster_hdbscan"].to_numpy() == -1
)

# Anggota asli masing-masing cluster
anggota_awal = {
    c: np.flatnonzero(
        df_rute["cluster_hdbscan"].to_numpy() == c
    )
    for c in cluster_awal
}

# Batas percobaan ukuran kelompok
maks_anggota = 9

# Dahulukan noise yang paling dekat dengan suatu cluster
urutan_noise = sorted(
    noise_idx,
    key=lambda i: min(
        D[i, anggota_awal[c]].mean()
        for c in cluster_awal
    )
)

df_rute["jarak_penugasan_m"] = np.nan

for i in urutan_noise:

    pilihan = [
        c for c in cluster_awal
        if (df_rute["kelompok_rute"] == c).sum() < maks_anggota
    ]

    if not pilihan:
        raise ValueError("Semua kelompok sudah mencapai batas anggota.")

    cluster_tujuan = min(
        pilihan,
        key=lambda c: D[i, anggota_awal[c]].mean()
    )

    jarak = D[i, anggota_awal[cluster_tujuan]].mean()

    df_rute.loc[i, "kelompok_rute"] = cluster_tujuan
    df_rute.loc[i, "jarak_penugasan_m"] = jarak

# Periksa hasil
print("Noise awal:", len(noise_idx))
print("Jumlah kelompok:", len(cluster_awal))
print("Belum mendapatkan kelompok:",
      (df_rute["kelompok_rute"] == -1).sum())

print("\nJumlah anggota tiap kelompok:")
display(
    df_rute["kelompok_rute"]
    .value_counts()
    .sort_index()
)

print("\nPenugasan puskesmas noise:")
display(
    df_rute.loc[
        df_rute["cluster_hdbscan"] == -1,
        [
            "Input Nama",
            "cluster_hdbscan",
            "kelompok_rute",
            "jarak_penugasan_m"
        ]
    ].sort_values("jarak_penugasan_m", ascending=False)
)

df_rute.to_csv(
    os.path.join(folder, "kelompok_distribusi.csv"),
    index=False
)
```

```python
cek_noise = []

for i in noise_idx:

    # Jarak rata-rata ke setiap cluster asli
    jarak_cluster = {}

    for c in cluster_awal:
        jarak_cluster[c] = D[i, anggota_awal[c]].mean()

    # Cluster yang benar-benar paling dekat
    terdekat = min(jarak_cluster, key=jarak_cluster.get)

    # Cluster yang dipilih kode sebelumnya
    dipilih = int(df_rute.loc[i, "kelompok_rute"])

    cek_noise.append({
        "puskesmas": df_rute.loc[i, "Input Nama"],
        "cluster_terdekat": terdekat,
        "cluster_dipilih": dipilih,
        "jarak_terdekat_km": round(jarak_cluster[terdekat] / 1000, 2),
        "jarak_dipilih_km": round(jarak_cluster[dipilih] / 1000, 2),
        "selisih_km": round(
            (jarak_cluster[dipilih] - jarak_cluster[terdekat]) / 1000, 2
        )
    })

df_cek = pd.DataFrame(cek_noise)

df_cek = df_cek.sort_values(
    "selisih_km",
    ascending=False
)

display(df_cek)

print(
    "Noise yang tidak masuk cluster terdekat:",
    (df_cek["cluster_terdekat"] != df_cek["cluster_dipilih"]).sum()
)
```
